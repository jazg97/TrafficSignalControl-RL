```python
import numpy as np
import matplotlib.pyplot as plt
import os
import pandas as pd
import math
```

```python
seed = 10
np.random.seed(seed)  # make tests reproducible
n_cars_generated = 1000
max_steps = 3600
# the generation of cars is distributed according to a weibull distribution
timings = np.random.weibull(2, n_cars_generated)
timings = np.sort(timings)

# reshape the distribution to fit the interval 0:max_steps
car_gen_steps = []
min_old = math.floor(timings[1])
max_old = math.ceil(timings[-1])
min_new = 0
max_new = max_steps
for value in timings:
    car_gen_steps = np.append(car_gen_steps, ((max_new - min_new) / (max_old - min_old)) * (value - max_old) + max_new)

weibull_car_gen_steps = np.rint(car_gen_steps)  # round every value to int -> effective steps when a car will be generated
```

```python
plt.figure()
plt.plot(car_gen_steps, np.linspace(1,1000,1000))
plt.xlabel('Time (s)')
plt.ylabel('Car (n)')
plt.show()
```

```python
shape = 10.0  # shape parameter for Pareto distribution

seed = 10
np.random.seed(seed)  # make tests reproducible
n_cars_generated = 1000
max_steps = 3600
# the generation of cars is distributed according to a weibull distribution
timings = np.random.pareto(shape, 1000)
timings = np.sort(timings)

# reshape the distribution to fit the interval 0:max_steps
car_gen_steps = []
min_old = math.floor(timings[1])
max_old = math.ceil(timings[-1])
min_new = 0
max_new = max_steps
for value in timings:
    car_gen_steps = np.append(car_gen_steps, ((max_new - min_new) / (max_old - min_old)) * (value - max_old) + max_new)

pareto_car_gen_steps = np.rint(car_gen_steps)  # round every value to int -> effective steps when a car will be generated
```

```python
plt.figure()
plt.plot(pareto_car_gen_steps, np.linspace(1,1000,1000))
plt.xlabel('Time (s)')
plt.ylabel('Car (n)')
plt.show()
```

```python
lam = 100  # average rate (lambda) for the Poisson distribution

seed = 10
np.random.seed(seed)  # make tests reproducible
n_cars_generated = 1000
max_steps = 3600
# the generation of cars is distributed according to a weibull distribution
timings = np.random.poisson(lam, 1000)
timings = np.sort(timings)

# reshape the distribution to fit the interval 0:max_steps
car_gen_steps = []
min_old = math.floor(timings[1])
max_old = math.ceil(timings[-1])
min_new = 0
max_new = max_steps
for value in timings:
    car_gen_steps = np.append(car_gen_steps, max(0, ((max_new - min_new) / (max_old - min_old)) * (value - max_old) + max_new))

poisson_car_gen_steps = np.rint(car_gen_steps)  # round every value to int -> effective steps when a car will be generated
```

```python
plt.figure()
plt.plot(poisson_car_gen_steps, np.linspace(1,1000,1000))
plt.xlabel('Time (s)')
plt.ylabel('Car (n)')
plt.show()
```

```python
plt.figure()
plt.plot(weibull_car_gen_steps, np.linspace(1,1000,1000), label='Weibull')
plt.plot(pareto_car_gen_steps, np.linspace(1,1000,1000), label='Pareto')
plt.plot(poisson_car_gen_steps, np.linspace(1,1000,1000), label='Poisson')
plt.xlabel('Time (s)')
plt.ylabel('Car (n)')
plt.legend()
plt.show()
```
